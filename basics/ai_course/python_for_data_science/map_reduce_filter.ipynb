```python
import math
radii = [1,2,3];
new_radii = list(map(lambda x : x * x * math.pi, radii))
print(new_radii)
```

```python
import math
data = [1,2,3,4,5,6,7,8,9]
new_data = list(map(lambda x : x * x * math.pi, data))
print(new_data)
```

```python
import statistics
data = [200,1,2,3,4,5,6,7,8,9]
avg = statistics.mean(data)
new_data = list(filter(lambda x : x > avg, data))
print(new_data)
```
