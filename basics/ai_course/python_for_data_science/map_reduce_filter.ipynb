```python
import math
radii = [1,2,3];
new_radii = list(map(lambda x : x * x * math.pi, radii))
print(new_radii)
```

```python
import math
data = [1,2,3,4,5,6,7,8,9]
new_data = list(map(lambda x : x * x * math.pi, data))
print(new_data)
```

```python
import statistics
data = [200,1,2,3,4,5,6,7,8,9]
avg = statistics.mean(data)
new_data = list(filter(lambda x : x > avg, data))
print(new_data)
```

```python
from functools import reduce
data = [1,2,3,4,5,5,6,7,8,9]
addition = reduce(lambda x, y : x + y , data)
print(addition)
```

```python
areas = [1,2,3,4,5,6,7,8,9]
new_areas = list(filter(lambda x : x > 5, areas))
print(new_areas)
```

```python
import re
name = ["Naveen", "Neha","", "Namarata"]
only_names = list(filter(lambda x : re.match(r"^[a-zA-Z]+$", x), name))
print(only_names)
```

```python
name = ["Naveen", "Neha","", "Namarata"]
only_names = list(filter(None, name))
print(only_names)
```

```python
import re
numbers = ["", "test", "summa", None]
only_none = list(filter(lambda x: x is not None and re.match("[a-zA-Z0-9]+", x), numbers))
print(only_none)
```

```python
from functools import reduce
data = [1,2,3,4,5]
print(reduce(lambda x, y : x * y, data))
```

```python
from functools import reduce

data = ["Naveen", "Prasanth", "is", "a", "good", "boy"]
print(reduce(lambda x, y : f"{x} {y}", data))
```
