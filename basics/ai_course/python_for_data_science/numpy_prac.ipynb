```python
import math

def round_half(n : float):
    return int(math.floor(n + 0.5))


values : list[float] = [2.5, 3.5, 4.5, 5.5]

rounded_half_up : list[float] = list(round_half(n=x) for x in values)

print(rounded_half_up)


avg_half_up : float = sum(rounded_half_up) / len(rounded_half_up)
print(avg_half_up)
```
